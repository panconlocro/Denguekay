Junta los parquets diarios cacheados, agrega a semanal y limpia nombres de distrito para poder cruzar después con socio/epi. Salida: `data/silver/meteo_semanal_distrital.csv`.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))
```

```python
import pandas as pd
from src.utils.paths import REFERENCE, SILVER
from src.processing.agregacion_semanal import (
    cargar_diario_desde_cache, agregar_a_semanal, limpiar_nombres_para_cruce,
)

distritos = pd.read_csv(REFERENCE / "distritos_piura_coords.csv")
```

```python
diario = cargar_diario_desde_cache(n_distritos_esperado=len(distritos))
diario.groupby("id_distrito").size().describe()
```

```python
semanal = agregar_a_semanal(diario, distritos)
semanal = limpiar_nombres_para_cruce(semanal)
semanal.head()
```

```python
SILVER.mkdir(parents=True, exist_ok=True)
semanal.to_csv(SILVER / "meteo_semanal_distrital.csv", index=False)
print(semanal.shape)
```
