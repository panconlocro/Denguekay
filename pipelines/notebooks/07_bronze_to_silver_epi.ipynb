# 07 — Bronze → Silver: epidemiológico

Cuenta notificaciones confirmadas (`tipo_dx = C`) de Piura por ubigeo, año y semana. Salida: `data/silver/epi_piura_semanal.csv`.

```python
from denguekay.config.paths import BRONZE_EPI_CSV, SILVER_EPI_HISTORICAL
from denguekay.data.processing.epi_cases import load_raw, build_weekly_cases

df_casos = build_weekly_cases(load_raw(BRONZE_EPI_CSV))
SILVER_EPI_HISTORICAL.parent.mkdir(parents=True, exist_ok=True)
df_casos.to_csv(SILVER_EPI_HISTORICAL, index=False)
print(f"{len(df_casos):,} distrito-semanas; {df_casos['casos_Dengue'].sum():,} casos confirmados.")
```
