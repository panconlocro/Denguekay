# 03b — Índice costero de El Niño (Niño 1+2, NOAA CPC)

Descarga `sstoi.indices` de NOAA CPC (OISST v2.1, base 1991-2020, mensual desde 1982) a `data/bronze/enso/`
con la fecha de descarga en el nombre (nunca se sobrescribe) y escribe `data/silver/enso_nino12_mensual.csv`
con el valor, la anomalía y `fecha_disponible` (día 10 del mes siguiente, regla conservadora según el README de
CPC). Fuente: https://www.cpc.ncep.noaa.gov/data/indices/sstoi.indices; dominio público
(https://www.weather.gov/disclaimer). Se corre junto con el clima de la 1b del estudio 06.


```python
from denguekay.config.paths import BRONZE_ENSO, SILVER_ENSO
from denguekay.data.ingestion.enso import download_sstoi
from denguekay.data.processing.enso_index import nino12_monthly

raw = download_sstoi()
raw
```

```python
nino = nino12_monthly(raw.read_text(encoding="utf-8"))
SILVER_ENSO.parent.mkdir(parents=True, exist_ok=True)
nino.to_csv(SILVER_ENSO, index=False)
print(nino.shape, nino.iloc[0].to_dict(), nino.iloc[-1].to_dict())
```
